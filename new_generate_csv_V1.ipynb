```python
import csv
import io
import zipfile
import logging
import pymysql
from datetime import datetime, timedelta

DB_USER = 'datasetgenerator'
DB_PASSWORD = 'yrCyuOxHUc3M88xHAyCMspJRwQR6jWjH'
DB_URL = '34.71.42.200'
DB_PORT = 3306
DB_SCHEMA = 'motionsensorsschema'

SYMMETRIC_WINDOW_SECONDS = 1

# Dataframes table rows
ROW_ID = 0
ROW_SENSORTYPEID = 1
ROW_ACTIVITYID = 2
ROW_ACTIVITYCONTEXTID = 3
ROW_SUBJECTID = 4
ROW_DEVICEID = 5
ROW_DEVICEBODYLOCATIONID = 6
ROW_RECEIVEDTIMESTAMP = 7
ROW_SAMPLINGRATEHZ = 8
ROW_TIMEWINDOWSECONDS = 9
ROW_STARTTIMESTAMP = 10
ROW_ENDTIMESTAMP = 11
ROW_READINGSNUMBER = 12
ROW_READINGSDATA = 13

# AggregateEventLogs table rows
AGGR_ROW_ID = 0
AGGR_ROW_AGGREGATE_ID = 1
AGGR_ROW_AGGREGATE_EVENT_ID = 2
AGGR_ROW_MAP_ID = 3
AGGR_ROW_SUBJECT_ID = 4
AGGR_ROW_REGISTEREDTIMESTAMP = 5
AGGR_ROW_RECEIVEDTIMESTAMP = 6

# Configure and create a logger
logging.basicConfig(level=logging.DEBUG,
                    format='%(asctime)s - %(levelname)s - %(message)s')

logger = logging.getLogger(__name__)
logger.setLevel(logging.INFO)

try:
    # Attempt to establish a connection to the database
    logger.info("Connecting to the MySQL database...")
    conn = pymysql.connect(
        user=DB_USER,
        password=DB_PASSWORD,
        host=DB_URL,
        port=DB_PORT,
        database=DB_SCHEMA,
        # PyMySQL doesn't use ssl_disabled parameter, instead use ssl parameter if needed
        # ssl={'ca': '/path/to/ca-cert.pem'} # Uncomment and modify if SSL connection is required
    )
    logger.info("Connection established.")

    cursor = conn.cursor()                                      # Create a cursor object using the cursor() method

    #query1 = 'SELECT * FROM dataframes ORDER BY StartTimeStamp' # Define the SQL query to select all data
    #query1 = 'SELECT * FROM dataframes WHERE SubjectID = "2" ORDER BY StartTimeStamp'
    #query1 = 'SELECT * FROM dataframes WHERE SubjectID = 2 and name = 'Tomasz' and surname='Artmanski' ORDER BY StartTimeStamp'
    query1 = 'SELECT * FROM motionsensorsschema.dataframes JOIN motionsensorsschema.subjects ON dataframes.SubjectId = subjects.Id WHERE SensorTypeId <= 8 AND Name = "Jonasz" AND Surname ="Kachel" ORDER BY StartTimestamp'
    #query1 = 'SELECT * FROM motionsensorsschema.dataframes JOIN motionsensorsschema.subjects ON dataframes.SubjectId = subjects.Id WHERE SensorTypeId <= 8 AND Name = "Olena" AND Surname ="Pavliuk" ORDER BY StartTimestamp'
    #query1 = 'SELECT * FROM motionsensorsschema.dataframes JOIN motionsensorsschema.subjects ON dataframes.SubjectId = subjects.Id WHERE SensorTypeId ==22 ORDER BY StartTimestamp'


    logger.info("Executing Dataframes select query...")
    cursor.execute(query1)                                      # Execute the SQL command
    results = cursor.fetchall()                                 # Fetch all the rows
    logger.info(f"Dataframes select query successful. Number of rows fetched: {len(results)}")
    query2 = 'SELECT * FROM motionsensorsschema.dataframes JOIN motionsensorsschema.subjects ON dataframes.SubjectId = subjects.Id WHERE SensorTypeId <= 8 AND Name = "Jonasz" AND Surname ="Kachel" ORDER BY StartTimestamp'

    #query2 = 'SELECT * FROM aggregateeventlogs ORDER BY RegisteredTimeStamp' # Define the SQL query to select all data
    #query2 = 'SELECT * FROM aggregateeventlogs WHERE SubjectID = "2" ORDER BY RegisteredTimeStamp'
    #query2 = 'SELECT * FROM aggregateeventlogs WHERE SubjectID = 2 ORDER BY RegisteredTimeStamp'
    #query2 = 'SELECT * FROM aggregateeventlogs WHERE SubjectID = 2 and name = "Tomasz" and surname="Artmanski" ORDER BY StartTimeStamp'
    #query2 = 'SELECT * FROM motionsensorsschema.dataframes JOIN motionsensorsschema.subjects WHERE AffiliationID = 2 ORDER BY StartTimeStamp'
    #query2 = 'SELECT * FROM motionsensorsschema.dataframes JOIN motionsensorsschema.subjects WHERE AffiliationID = 2 AND Name = "Kamila" AND Surname ="Mlynarczyk" ORDER BY StartTimeStamp'
    #query2 = 'SELECT * FROM motionsensorsschema.dataframes JOIN motionsensorsschema.subjects ON dataframes.SubjectId = subjects.Id WHERE SensorTypeId <= 8 AND Name = "Olena" AND Surname ="Pavliuk" ORDER BY StartTimestamp'
    #query2 = 'SELECT * FROM motionsensorsschema.dataframes JOIN motionsensorsschema.subjects ON dataframes.SubjectId = subjects.Id WHERE SensorTypeId ==22 ORDER BY StartTimestamp'
    #query2 = 'SELECT * FROM dataframes ORDER BY StartTimeStamp'
    #query2 = 'SELECT * FROM motionsensorsschema.dataframes JOIN motionsensorsschema.subjects WHERE AffiliationID = 2 AND Name = "Maja" AND Surname ="Liszczyk" ORDER BY StartTimeStamp'

    logger.info("Executing AggregateEventLogs select query...")
    cursor.execute(query2)                                          # Execute the SQL command
    aggr_results = cursor.fetchall()                                # Fetch all the rows
    logger.info(f"AggregateEventLogs select query. Number of rows fetched: {len(aggr_results)}")

except pymysql.Error as e:
    logger.error(f"SQL error occurred: {e}")
except Exception as e:
    logger.error(f"General error occurred: {e}")
finally:
    # Close the cursor and connection
    if 'cursor' in locals():
        cursor.close()
        logger.info("Cursor closed.")
    if 'conn' in locals():
        try:
            # PyMySQL uses a different method to check connection status
            conn.close()
            logger.info("Connection closed.")
        except:
            pass

#######################################
#           Grouping                  #
#######################################

logger.info(f"Grouping frames by start timestamps with symmetric window {SYMMETRIC_WINDOW_SECONDS}")

# Group the readings by rounded start timestamps within the SYMMETRIC_WINDOW_SECONDS
grouped_readings = {}

def round_time_and_add_seconds(ts, sec):
    return ts.replace(microsecond=0) + timedelta(seconds=sec)

for row in results:
    start_timestamp = row[ROW_STARTTIMESTAMP]
    new_group = True

    for i in range(-SYMMETRIC_WINDOW_SECONDS, SYMMETRIC_WINDOW_SECONDS + 1):
        key_ts = round_time_and_add_seconds(start_timestamp, i)
        if key_ts in grouped_readings:
            new_group = False
            grouped_readings[key_ts].append(row)
            break

    if (new_group):
        key_ts = round_time_and_add_seconds(start_timestamp, 0)
        grouped_readings[key_ts] = []
        grouped_readings[key_ts].append(row)

logger.info(f"Found groups {len(grouped_readings)}")

#######################################
#      Validating and sorting         #
#######################################

logger.info("Processing groups")

def sort_key(row):
    return row[ROW_SENSORTYPEID]

def print_group(group):
    res = ""
    for row in group:
        res += f"id={row[ROW_ID]}, activityId={row[ROW_ACTIVITYID]}, sensorTypeId={row[ROW_SENSORTYPEID]}, " + \
               f"startTimestamp={row[ROW_STARTTIMESTAMP]}\n"
    return res

for key_ts in list(grouped_readings.keys()):  # Create a list of keys to iterate over
    group = grouped_readings[key_ts]
    group_size = len(group)

    if group_size == 6:
        # Sort the group in place by the second element of each row
        group.sort(key=sort_key)

        # Check if the sorted values are 3, 4, 5, 6, 7, 8
        if not [sort_key(row) for row in group] == [3, 4, 5, 6, 7, 8]:
            removed_group = grouped_readings.pop(key_ts)
            logger.warning(f"Group for {key_ts} does not have the correct sorted values:\n{print_group(removed_group)}")

    else:
        # Remove the key from the dictionary
        removed_group = grouped_readings.pop(key_ts)

        # Log the type of group and its content
        if group_size < 6:
            logger.warning(f"Incomplete group size <{group_size}> for {key_ts}:\n{print_group(removed_group)}")
        else:
            logger.warning(f"Overflown group size <{group_size}> for {key_ts}:\n{print_group(removed_group)}")

logger.info(f"Successfully processed groups: {len(grouped_readings)}\n")

#############################################
#  Generating the CSV files for dataframes  #
#############################################

logger.info("Generating the CSV files for dataframes")

# Create in-memory file-like objects for each category
output_3s = io.StringIO()
writer_3s = csv.writer(output_3s)

output_6s = io.StringIO()
writer_6s = csv.writer(output_6s)

# Iterate over each group
for group in grouped_readings.items():
    group_data = [
        group[1][0][ROW_STARTTIMESTAMP],        # start timestamp for the first frame
        group[1][0][ROW_ENDTIMESTAMP],          # end timestamp for the first frame
        group[1][0][ROW_ACTIVITYID],            # activityId for the first frame
        group[1][0][ROW_DEVICEBODYLOCATIONID],  # devicebodylocationId for the first frame
        group[1][0][ROW_SUBJECTID],             # subjectId for the first frame
    ]

    # Append ReadingsData from all 6 frames
    for row in group[1]:                        # readings for each frame
        group_data.extend(row[ROW_READINGSDATA].split(','))

    # Write the row to the appropriate file-like object
    if group[1][0][ROW_TIMEWINDOWSECONDS] == 6:
        writer_6s.writerow(group_data)
    elif group[1][0][ROW_TIMEWINDOWSECONDS] == 3:
        writer_3s.writerow(group_data)
    else:
        logger.warn(f"Unsupported time window size {group[1][0][ROW_TIMEWINDOWSECONDS]}")

logger.info(f"Successfully generated the CSV files for Dataframes")

#############################################
#  Generating the CSV files for aggregates  #
#############################################

logger.info("Generating the CSV files for aggregates")

# Create in-memory file-like objects for aggregates data
output_aggr = io.StringIO()
writer_aggr = csv.writer(output_aggr)

# Define column headers
headers = [
    "RegisteredTimestamp",
    "AggregateID",
    "AggregateEventID",
    "MapID",
    "SubjectID"
]

# Write the header row
writer_aggr.writerow(headers)

for aggregate in aggr_results:
    aggregate_data = [
        aggregate[AGGR_ROW_REGISTEREDTIMESTAMP],
        aggregate[AGGR_ROW_AGGREGATE_ID] ,
        aggregate[AGGR_ROW_AGGREGATE_EVENT_ID],
        aggregate[AGGR_ROW_MAP_ID],
        aggregate[AGGR_ROW_SUBJECT_ID],
    ]

    # Write the row to the appropriate file-like object
    writer_aggr.writerow(aggregate_data)

logger.info(f"Successfully generated the CSV files for aggregates")

#############################################
#       Generating the Dataset archive      #
#############################################

logger.info(f"Generating the Dataset archive")

# Function to add data to the ZIP archive
def add_to_zip(zip_file, file_name, data):
    with zip_file.open(file_name, mode='w') as file:
        file.write(data.encode('utf-8'))

# Get the current datetime
current_datetime = datetime.now().strftime("%Y%m%d_%H%M%S")

# Create the filenames
zip_file_name = f'dataset_{current_datetime}.zip'
csv_file_name_6s = f"dataframes_6s_{current_datetime}.csv"
csv_file_name_3s = f"dataframes_3s_{current_datetime}.csv"
csv_file_name_aggr = f"aggregates_{current_datetime}.csv"

with zipfile.ZipFile(zip_file_name, mode='w') as zipf:
    #add_to_zip(zipf, csv_file_name_6s, output_6s.getvalue())
    add_to_zip(zipf, csv_file_name_3s, output_3s.getvalue())
    #add_to_zip(zipf, csv_file_name_aggr, output_aggr.getvalue())

logger.info(f"ZIP file created at {zip_file_name}")
```
